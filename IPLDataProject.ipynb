# Custom Questions & Insights

```python
# Who won the highest margin by runs?


df[df['won_by'] == 'Runs'].sort_values(by = 'margin',ascending=False).head(1)[['match_winner','margin']]
```

```python
# Which player had the highest individual score?


df[df['highscore'] == df['highscore'].max()][['top_scorer','highscore']]
```

```python
# Ensure 'highscore' is numeric
df['highscore'] = pd.to_numeric(df['highscore'], errors='coerce')

# Find the row with the highest score
top_scorer_row = df.loc[df['highscore'].idxmax()]

# Extract player name and score
top_scorer = top_scorer_row['top_scorer']
top_score = top_scorer_row['highscore']

print(f"The highest individual score is {top_score} by {top_scorer}.")
```

```python
# Which bowler had the best bowling figures?


df[df['highest_wickets'] == df['highest_wickets'].max()][['best_bowling','best_bowling_figure']]
```
